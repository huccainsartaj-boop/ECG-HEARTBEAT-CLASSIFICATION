```python
import os
import kagglehub
from pathlib import Path

path = kagglehub.dataset_download("shayanfazeli/heartbeat")

print("Dataset download location:", path)

# Dataset ke andar mojood CSV files search karein
csv_files = list(Path(path).rglob("*.csv"))

print("\nCSV files found:", len(csv_files))

for file in csv_files:
    print(file)
```

```python
# CSV files ko filename ke through identify karein
csv_map = {
    file.name.lower(): str(file)
    for file in csv_files
}

print("Available CSV filenames:")
for name in csv_map:
    print(name)

train_candidates = [
    name for name in csv_map
    if "mitbih_train" in name
]

test_candidates = [
    name for name in csv_map
    if "mitbih_test" in name
]

if not train_candidates or not test_candidates:
    raise FileNotFoundError(
        "MIT-BIH train/test CSV files nahi milin. "
        "Cell 2 mein print hone wale filenames check karein."
    )

TRAIN_CSV = csv_map[train_candidates[0]]
TEST_CSV = csv_map[test_candidates[0]]

print("\nTraining CSV:", TRAIN_CSV)
print("Testing CSV:", TEST_CSV)
```

```python
import pandas as pd
import numpy as np

train_df = pd.read_csv(TRAIN_CSV, header=None)
test_df = pd.read_csv(TEST_CSV, header=None)

print("Training dataset shape:", train_df.shape)
print("Testing dataset shape:", test_df.shape)

print("\nFirst 5 training rows:")
display(train_df.head())

print("\nFirst 5 testing rows:")
display(test_df.head())
```

```python
print("Training rows:", len(train_df))
print("Testing rows:", len(test_df))

print("Total columns:", train_df.shape[1])
print("Missing values in training:", train_df.isnull().sum().sum())
print("Missing values in testing:", test_df.isnull().sum().sum())

print("\nData types:")
print(train_df.dtypes.value_counts())

print("\nLast column: class labels")
print(train_df.iloc[:, -1].value_counts().sort_index())
```

```python
# Last column class label hai
X_train_full = train_df.iloc[:, :-1].to_numpy(dtype=np.float32)
y_train_full = train_df.iloc[:, -1].to_numpy()

X_test = test_df.iloc[:, :-1].to_numpy(dtype=np.float32)
y_test = test_df.iloc[:, -1].to_numpy()

# Labels ko integer mein convert karein
y_train_full = y_train_full.astype(np.int32)
y_test = y_test.astype(np.int32)

print("Training signals:", X_train_full.shape)
print("Training labels:", y_train_full.shape)

print("Testing signals:", X_test.shape)
print("Testing labels:", y_test.shape)

print("Training classes:", np.unique(y_train_full))
print("Testing classes:", np.unique(y_test))
```

```python
# Invalid values ko NaN mein convert karein
X_train_full = np.where(
    np.isfinite(X_train_full), X_train_full, np.nan
).astype(np.float32)

X_test = np.where(
    np.isfinite(X_test), X_test, np.nan
).astype(np.float32)

# Har signal column ke liye training data ka median
medians = np.nanmedian(X_train_full, axis=0)
medians = np.nan_to_num(medians, nan=0.0)

# Train aur test dono mein wahi training medians use karein
for j in range(X_train_full.shape[1]):
    X_train_full[np.isnan(X_train_full[:, j]), j] = medians[j]
    X_test[np.isnan(X_test[:, j]), j] = medians[j]

print("Remaining missing values:", np.isnan(X_train_full).sum())
print("Remaining invalid values:", np.isfinite(X_test).all())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

class_counts = pd.Series(y_train_full).value_counts().sort_index()

plt.figure(figsize=(9, 5))
sns.barplot(x=class_counts.index, y=class_counts.values)
plt.title("ECG Heartbeat Class Distribution")
plt.xlabel("Heartbeat Class")
plt.ylabel("Number of Samples")
plt.show()

print(class_counts)
```

```python
class_names = {
    0: "Normal",
    1: "Supraventricular",
    2: "Ventricular",
    3: "Fusion",
    4: "Unknown"
}

plt.figure(figsize=(15, 10))

for label in sorted(np.unique(y_train_full)):
    indices = np.where(y_train_full == label)[0]

    if len(indices) == 0:
        continue

    sample_index = indices[0]

    plt.subplot(3, 2, int(label) + 1)
    plt.plot(X_train_full[sample_index])
    plt.title(f"Class {label}: {class_names.get(int(label), 'Other')}")
    plt.xlabel("Time Step")
    plt.ylabel("Signal Amplitude")
    plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_val, y_train, y_val = train_test_split(
    X_train_full,
    y_train_full,
    test_size=0.15,
    random_state=42,
    stratify=y_train_full
)

print("Training:", X_train.shape, y_train.shape)
print("Validation:", X_val.shape, y_val.shape)
print("Testing:", X_test.shape, y_test.shape)
```

```python
X_train_rnn = X_train[..., np.newaxis]
X_val_rnn = X_val[..., np.newaxis]
X_test_rnn = X_test[..., np.newaxis]

print("RNN training shape:", X_train_rnn.shape)
print("RNN validation shape:", X_val_rnn.shape)
print("RNN testing shape:", X_test_rnn.shape)
```

```python
from sklearn.utils.class_weight import compute_class_weight

classes = np.unique(y_train)

weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=y_train
)

class_weights = {
    int(cls): float(weight)
    for cls, weight in zip(classes, weights)
}

print("Class weights:")
print(class_weights)
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models, regularizers

tf.keras.utils.set_random_seed(42)

model = models.Sequential([
    layers.Input(shape=(187, 1)),

    layers.Bidirectional(
        layers.GRU(
            64,
            return_sequences=True,
            dropout=0.15
        )
    ),

    layers.LayerNormalization(),

    layers.LSTM(
        32,
        dropout=0.20,
        kernel_regularizer=regularizers.l2(1e-4)
    ),

    layers.Dense(
        64,
        activation="relu",
        kernel_regularizer=regularizers.l2(1e-4)
    ),

    layers.Dropout(0.30),

    layers.Dense(5, activation="softmax")
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python
from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau,
    ModelCheckpoint
)

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=6,
    restore_best_weights=True,
    verbose=1
)

reduce_lr = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=2,
    min_lr=1e-6,
    verbose=1
)

checkpoint = ModelCheckpoint(
    "/content/best_heartbeat_rnn.keras",
    monitor="val_loss",
    save_best_only=True,
    verbose=1
)

print("Callbacks ready.")
```

```python
EPOCHS = 50
BATCH_SIZE = 256

history = model.fit(
    X_train_rnn,
    y_train,
    validation_data=(X_val_rnn, y_val),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    class_weight=class_weights,
    callbacks=[
        early_stopping,
        reduce_lr,
        checkpoint
    ],
    verbose=1
)
```

```python
model = tf.keras.models.load_model(
    "/content/best_heartbeat_rnn.keras"
)

test_loss, test_accuracy = model.evaluate(
    X_test_rnn,
    y_test,
    batch_size=256,
    verbose=1
)

print(f"Improved Test Loss: {test_loss:.4f}")
print(f"Improved Test Accuracy: {test_accuracy * 100:.2f}%")
```

```python
plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.plot(history.history["accuracy"], label="Training Accuracy")
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")
plt.title("RNN Model Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.title("RNN Model Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models, regularizers

tf.keras.utils.set_random_seed(42)

model = models.Sequential([
    layers.Input(shape=(187, 1)),

    # ECG signal feature extraction
    layers.Conv1D(
        filters=64,
        kernel_size=5,
        padding="same",
        activation="relu"
    ),
    layers.BatchNormalization(),
    layers.MaxPooling1D(pool_size=2),

    # Bidirectional GRU
    layers.Bidirectional(
        layers.GRU(64, return_sequences=True)
    ),
    layers.Dropout(0.25),

    # Bidirectional LSTM
    layers.Bidirectional(
        layers.LSTM(32)
    ),
    layers.BatchNormalization(),

    layers.Dense(
        64,
        activation="relu",
        kernel_regularizer=regularizers.l2(1e-4)
    ),
    layers.Dropout(0.35),

    layers.Dense(5, activation="softmax")
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.0005
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python
from sklearn.metrics import classification_report

y_prob = model.predict(
    X_test_rnn,
    batch_size=BATCH_SIZE,
    verbose=1
)

y_pred = np.argmax(y_prob, axis=1)

labels = [0, 1, 2, 3, 4]
target_names = [
    "Normal",
    "Supraventricular",
    "Ventricular",
    "Fusion",
    "Unknown"
]

print(classification_report(
    y_test,
    y_pred,
    labels=labels,
    target_names=target_names,
    digits=4,
    zero_division=0
))
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred, labels=labels)

plt.figure(figsize=(9, 7))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=target_names,
    yticklabels=target_names
)

plt.title("ECG Heartbeat RNN - Confusion Matrix")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.xticks(rotation=30, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
sample_index = 0

sample = X_test_rnn[sample_index:sample_index + 1]

probabilities = model.predict(sample, verbose=0)[0]

predicted_class = int(np.argmax(probabilities))
actual_class = int(y_test[sample_index])

plt.figure(figsize=(12, 4))
plt.plot(X_test[sample_index])
plt.title("ECG Heartbeat Signal")
plt.xlabel("Time Step")
plt.ylabel("Signal Amplitude")
plt.grid(True, alpha=0.3)
plt.show()

print("Actual class:", class_names.get(actual_class, actual_class))
print("Predicted class:", class_names.get(predicted_class, predicted_class))
print(f"Model confidence: {probabilities[predicted_class] * 100:.2f}%")

print("\nClass probabilities:")
for i, prob in enumerate(probabilities):
    print(f"{class_names[i]}: {prob * 100:.2f}%")
```
